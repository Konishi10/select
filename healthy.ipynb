```python
import os
import shutil
import pandas as pd


# =========================
# 1. CSVから正常画像のみ抽出
# =========================

# CSVを読み込む
df = pd.read_csv("train_1.csv")

# diagnosis=0（正常）のみ抽出
healthy_df = df[df["diagnosis"] == 0].copy()

print(f"正常画像数: {len(healthy_df)}")

# 正常画像のみのCSVを保存
healthy_df.to_csv("healthy_images.csv", index=False)

print("healthy_images.csv を保存しました")


# =========================
# 2. 正常画像のみコピー
# =========================

csv_path = "healthy_images.csv"

# ここが重要
# 画像は train_images/train_images の中にある
image_dir = "train_images/train_images"

output_dir = "healthy_images"

# 出力フォルダを作成
os.makedirs(output_dir, exist_ok=True)

# healthy_images.csv を読み込む
df = pd.read_csv(csv_path)

copied_count = 0
missing_count = 0

for img_id in df["id_code"]:
    src = os.path.join(image_dir, img_id + ".png")
    dst = os.path.join(output_dir, img_id + ".png")

    if os.path.exists(src):
        shutil.copy(src, dst)
        copied_count += 1
    else:
        print(f"画像が見つかりません: {src}")
        missing_count += 1

print("正常画像のみコピー完了")
print(f"コピーできた画像数: {copied_count}")
print(f"見つからなかった画像数: {missing_count}")
```
