```python
import numpy as np

def vessel_distribution_score(
    vessel_skeleton,
    grid_size=8
):
    """
    血管が画像全体にどれだけ分布しているかを評価する。

    Parameters
    ----------
    vessel_skeleton : ndarray
        細線化済み血管画像（0/255）

    grid_size : int
        グリッド分割数

    Returns
    -------
    score : float
        0～1
        大きいほど画像全体に血管が分布
    """

    binary = vessel_skeleton > 0

    h, w = binary.shape

    cell_h = h // grid_size
    cell_w = w // grid_size

    occupied_cells = 0
    total_cells = grid_size * grid_size

    for gy in range(grid_size):
        for gx in range(grid_size):

            y0 = gy * cell_h
            y1 = h if gy == grid_size - 1 else (gy + 1) * cell_h

            x0 = gx * cell_w
            x1 = w if gx == grid_size - 1 else (gx + 1) * cell_w

            cell = binary[y0:y1, x0:x1]

            if np.any(cell):
                occupied_cells += 1

    score = occupied_cells / total_cells

    return score
```
